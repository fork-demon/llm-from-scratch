# Module 07: Self-Attention (Interactive)

Explore how attention routes information using Queries, Keys, and Values. We'll visualize the attention weight matrix (the "heatmap") to see exactly which tokens are looking at which other tokens.

```python
import numpy as np
import matplotlib.pyplot as plt
from attention_numpy import attention, MultiHeadAttention

%matplotlib inline
np.random.seed(42)
```

## 1. Visualizing the Causal Attention Mask
We create a dummy sequence and visualize the attention weights. Causal masking ensures tokens can only look at themselves and the past (left side of the matrix), never the future.

```python
T, D = 8, 16  # 8 tokens, 16-dim embeddings
X = np.random.randn(T, D)

Wq = np.random.randn(D, D) / np.sqrt(D)
Wk = np.random.randn(D, D) / np.sqrt(D)
Wv = np.random.randn(D, D) / np.sqrt(D)

# Run attention with causal=True
out, weights = attention(X, Wq, Wk, Wv, causal=True)

plt.figure(figsize=(6, 5))
plt.imshow(weights, cmap='viridis')
plt.colorbar(label='Attention Weight (Softmax)')
plt.title("Causal Attention Weights (Lower Triangular)")
plt.xlabel("Key Position (Source)")
plt.ylabel("Query Position (Target)")
plt.show()
```

## 2. Testing Multi-Head Attention Shapes
Multi-head attention splits the embedding dimension `D` across `H` heads, runs attention in parallel, and concatenates the results back together. Notice how the input and output shapes match perfectly.

```python
mha = MultiHeadAttention(d_model=16, num_heads=4)
out = mha.forward(X)
print(f"Input shape:  {X.shape}  [Time, Dim]")
print(f"Output shape: {out.shape}  [Time, Dim]")
assert X.shape == out.shape, "MHA broke the shape!"
print("\nShape invariant holds. The transformer block can now cleanly add this output to the residual stream.")
```
