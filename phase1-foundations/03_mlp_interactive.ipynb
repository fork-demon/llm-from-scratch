Welcome to the interactive companion notebook for Module 03!
In this notebook, you will:
1. Synthesize a 3-class spiral dataset.
2. Train a linear model and see its performance ceiling.
3. Build a Multi-Layer Perceptron (MLP) with ReLU activations from scratch in NumPy.
4. Plot the decision boundaries interactively.
5. Verify backprop against a numerical finite-difference gradient check.

```python
import numpy as np
import matplotlib.pyplot as plt
from mlp_numpy import make_spiral, train_linear, train_mlp, gradient_check, plot_boundaries, MLP

%matplotlib inline
print("Libraries loaded successfully.")
```

## 1. Generate & Inspect the Spiral Dataset
The spiral problem is non-linearly separable in 2D space. A straight linear boundary cannot partition the three spiral arms.

```python
X, y = make_spiral(points_per_class=100, num_classes=3)
print(f"X shape: {X.shape}, y shape: {y.shape}")

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Spectral, edgecolors='k', s=25)
plt.title("3-Class Spiral Dataset")
plt.xlabel("x0")
plt.ylabel("x1")
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Train a Linear Classifier (Watch It Stall)
Watch the accuracy plateau around ~50-54%. Stacking more linear layers without non-linearities would mathematically collapse back to this exact same line.

```python
loss, acc, W_lin, b_lin = train_linear(X, y, steps=300, lr=1.0)
print(f"Linear Model -> Final Loss: {loss:.4f}, Accuracy: {acc:.3f}")
```

## 3. Train the 2-Hidden-Layer MLP (ReLU Hinges Bend Space)
By adding `ReLU(x) = max(0, x)`, the network can compose piece-wise linear bends to carve complex non-linear boundaries.

```python
net = train_mlp(X, y, steps=2000, lr=0.5)
print(f"MLP Model -> Final Accuracy: {net.accuracy(X, y):.3f}")
```

## 4. Visualizing Decision Boundaries Side-by-Side
Compare how each model partitions the 2D input plane.

```python
plot_boundaries(
    X, y,
    linear_predict_fn=lambda g: (g @ W_lin + b_lin).argmax(axis=1),
    mlp_predict_fn=lambda g: net.forward(g).argmax(axis=1)
)
```

## 5. Gradient Check Verification
Prove your backprop analytical gradients against brute-force centered differences:
$$\frac{\partial L}{\partial W} \approx \frac{L(W + h) - L(W - h)}{2h}$$

```python
gradient_check(net, X[:50], y[:50])
```
