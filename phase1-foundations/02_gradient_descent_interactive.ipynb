# Module 02: Gradient Descent

This notebook provides a visual, interactive way to explore gradient descent. You will:
1. Fit a linear model using numerical gradients (the "nudge" method).
2. Visualize the loss landscape and the trajectory of the weights.

```python
import numpy as np
import matplotlib.pyplot as plt
from gradient_descent import train_linear, loss_fn

%matplotlib inline
rng = np.random.default_rng(42)
```

## 1. Generate Toy Data (Linear)
We create a dataset following `y = 3x + 2` with some noise.

```python
X = np.random.uniform(-2, 2, 100)
Y = 3 * X + 2 + np.random.normal(0, 0.5, 100)

plt.scatter(X, Y, alpha=0.6)
plt.title("Toy Dataset: y = 3x + 2 + noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## 2. Train and Visualize Trajectory
We will track the weights `(w, b)` as they descend the loss valley.

```python
history = []
w, b = 0.0, 0.0  # Start wrong
lr = 0.1
h = 1e-4

for step in range(50):
    current_loss = loss_fn(w, b, X, Y)
    history.append((w, b, current_loss))
    
    # Numerical gradients (the nudge method)
    loss_w_up = loss_fn(w + h, b, X, Y)
    grad_w = (loss_w_up - current_loss) / h
    
    loss_b_up = loss_fn(w, b + h, X, Y)
    grad_b = (loss_b_up - current_loss) / h
    
    w -= lr * grad_w
    b -= lr * grad_b

print(f"Final w: {w:.3f} (target 3.0), Final b: {b:.3f} (target 2.0)")

ws = [h[0] for h in history]
bs = [h[1] for h in history]
losses = [h[2] for h in history]

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(losses, marker='o', markersize=4)
plt.title("Loss over Time")
plt.xlabel("Step")
plt.ylabel("MSE Loss")

plt.subplot(1, 2, 2)
plt.plot(ws, bs, marker='o', markersize=4, color='red')
plt.scatter([3.0], [2.0], color='green', marker='*', s=200, label="True Target (3, 2)")
plt.title("Weight Trajectory in (w, b) Space")
plt.xlabel("Weight (w)")
plt.ylabel("Bias (b)")
plt.legend()
plt.tight_layout()
plt.show()
```
